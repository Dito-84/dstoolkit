#### Creating a Sample Dataset
###### Note: Using this method creates a list

```python
sample_dataset = [72,37,56,41,22,76,35,11,19,24,77,84,6,32,44,52]
print(sample_dataset)
```

#### Importing numpy & using it to transform data into an array

###### Note: This allows us to perform mathematical operations on the items within the "list"

```python
import numpy as np

grades = np.array(sample_dataset)
print(grades)

grades.shape
```

```python
# Single item 
#grades[5]

# Multiple items
print(grades[[5,6,7]])
```

```python
grades.mean()
```

#### Finding the lowest grade in the dataset

```python
grades.min()
```

#### Finding the highest grade in the dataset

#### Creating a second dataset to represent hours studied

```python
study_hours = [10,4,8,5,3,11,4,2,3,3,9,12,1,4,5,7]
print(study_hours)
```

#### Using numpy to create a new array called 'student_data' containing 2 arrays: 'grades' and 'study_hours'

```python
student_data = np.array([study_hours,grades])
```

#### Syntax for extracting individual items from the arrays within the 'parent' array

```python
#The array index begins at '0'. In this example, this represents 'study_hours' 
student_data[0][1]
```

#### Finding the mean StudyHours and mean Grade/Creating Variables

```python
avg_study = student_data[0].mean()
avg_grade = student_data[1].mean()

# Using formatted string to present averages
print(f'Average study time: {avg_study}\nAverage grade: {avg_grade}')
```

#### Reading a CSV file into a dataframe
###### In this example, the file is in the same directory folder, so we can use the file name 

```python
df_students_2 = pd.read_csv('../data/StudentData.csv')
print(df_students_2)
```

#### Identifying null values
###### The isnull() method returns a DataFrame objectwhere all the values are replaced with a Boolean value True for NULL values, and otherwise False.

```python
df_students_2.isnull()
```

#### Finding the total number of null vales in each column

```python
df_students_2.isnull().sum()
```

#### Using the dropna( ) function to remove rows that contain NULL values.

###### The dropna() method returns a new DataFrame object unless the inplace parameter is set to True, in that case the dropna() method does the removing in the original DataFrame instead.

```python
df_students_2 = df_students_2.dropna(axis=0) # Rows are axis 0 and columns are axis 1

df_students_2.isnull().sum()
```

#### Finding the mean study_hours and Grade using column names as an index

```python
mean_study = df_students_2["StudyHours"].mean()

mean_grade = df_students_2["Grade"].mean()

print(mean_study, mean_grade)
```

#### Filtering the dataframe to find only the students who studied more than the mean study_hours

```python
df_students_2[df_students_2.StudyHours > mean_study]
```

#### Finding the mean grade for students who studied more than the mean study_hours

```python
df_students_2[df_students_2.StudyHours > mean_study].Grade.mean()
```

#### Create a new variable to reflect whether a student passed using the pandas series( ) function & adding the series to the dataframe using the concat( ) function
###### Note: this will generate and add a column which contains 'True' if the student passed and 'False' if they did not


```python
passed = pd.Series(df_students_2.Grade >= 60)
df_students_2 = pd.concat([df_students_2, passed.rename('Passed')], axis = 1)
```

#### Using the groupby( ) operation to separate students into two goups according to whether they passed

```python
df_students_2.groupby(df_students_2.Passed).Name.count()
```

#### Finding/Printing the mean grade for each group

```python
print(df_students_2.groupby(df_students_2.Passed).Grade.mean())
```

#### Finding/Printing the mean Grade & StudyHours for each group

```python
print(df_students_2.groupby(df_students_2.Passed)[["Grade", "StudyHours"]].mean())
```

#### Sorting the dataframe in ascending order according to grade

```python
df_students_2 = df_students_2.sort_values('Grade')
# Alternative Syntax for sorting in decending order
#df_students_2 = df_students_2.sort_values('Grade', ascending=False)
```
