```python
import  pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv('StudentData.csv')
```

```python
print(df)
```

```python
def drop_column_null(var_data):
    # Always return the data: null rows removed, or an unchanged copy if there are none
    return var_data.dropna(axis = 0)

drop_column_null(df.Grade)
```

```python
def show_histogram(var_data):

    clean_var_data = var_data.dropna(axis = 0, inplace = False)
    
    fig = plt.figure(figsize=(10,4))

    plt.hist(clean_var_data)

    plt.axvline(x=clean_var_data.mean(), color='cyan', linestyle = 'dashed', linewidth = 2)
    plt.axvline(x=clean_var_data.median(), color='red', linestyle = 'dashed', linewidth = 2)
    plt.axvline(x=clean_var_data.mode()[0], color='yellow', linestyle = 'dashed', linewidth = 2)

    plt.title("Data Distribution")
    plt.xlabel("Value")
    plt.ylabel("Frequency")

    plt.show()

show_histogram(df.StudyHours)
```

```python
def show_distribution(var_data):

    clean_var_data = var_data.dropna(axis = 0, inplace = False)

    fig, ax = plt.subplots(2,1, figsize=(12.8, 7.2))

    ax[0].hist(clean_var_data)
    ax[0].set_ylabel("Frequency")

    ax[0].axvline(x=clean_var_data.mean(), color='cyan', linestyle = 'dashed', linewidth = 2)
    ax[0].axvline(x=clean_var_data.median(), color='red', linestyle = 'dashed', linewidth = 2)
    ax[0].axvline(x=clean_var_data.mode()[0], color='yellow', linestyle = 'dashed', linewidth = 2)

    ax[1].boxplot(clean_var_data, vert=False)

    ax[1].set_xlabel('Value')

    fig.suptitle("Data Distribution")

    plt.show()

show_distribution(df.Grade)
```

```python
def show_distribution_profile(var_data3):

    fig, ax = plt.subplots(2,2, figsize=(12.8, 7.2))

    ax[0, 0].hist(var_data3)
    ax[0, 0].set_ylabel("Frequency")

    ax[0, 0].axvline(x=var_data3.mean(), color='cyan', linestyle = 'dashed', linewidth = 2)
    ax[0, 0].axvline(x=var_data3.median(), color='green', linestyle = 'dashed', linewidth = 2)

    var_data3.plot.density(ax = ax[0,1])

    var_data3.plot.box(ax= ax[1,0], vert=False)

    ax[1,1].axis('off')
    table = pd.plotting.table(ax[1,1], var_data3.describe().round(2), loc='center', cellLoc='center', colWidths=list([.4, .4]))

    fig.suptitle("Data Distribution")

    plt.subplots_adjust(wspace=0.25)

    plt.show()
show_distribution_profile(df.Grade)
```

```python
show_distribution_profile(df.StudyHours)
```

#### Using the reusable versions in `toolkit.py`
###### The functions above are collected in `toolkit.py` (with null handling built in) so any notebook can import them

```python
import toolkit

toolkit.show_eda(df.Grade)
toolkit.show_distribution_profile(df.Grade, title="Grade Distribution")

m, b, predict = toolkit.fit_line(df.StudyHours, df.Grade)
print(f"Grade = {m:.2f} * StudyHours + {b:.2f}")
print("Predicted grade for 9 hours:", round(predict(9), 1))
```
